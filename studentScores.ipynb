>> ## Plot a scatter plot of Hours vs. Scores — is there a visible relationship?

```python
plt.scatter(df['Hours'], df['Scores'])
plt.xlabel('Study Hours')
plt.ylabel('Scores')
plt.title('Student Scores')
plt.grid(True)
plt.show()
```
